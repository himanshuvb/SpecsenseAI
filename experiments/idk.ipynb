```python

import os
import pandas as pd

json_file_path = os.path.join("..", "inventory", "inventory.json")

inventory = pd.read_json(json_file_path)   

inventory.head()
```

```python
print(inventory.head())
print(inventory.columns)
print(inventory.dtypes)
```

```python
specs = pd.json_normalize(inventory["specs"])
print(specs)
inventory = pd.concat(
    [inventory.drop(columns=["specs"]), specs],
    axis=1
)
print(inventory.head())
```

```python
user_requirements = {
    "weight_kg": 1.5,
    "ram_gb": 8,
    "storage_gb": 256,
    "battery_hours": 8
}
```

```python
matching_products = inventory[
    (inventory["weight_kg"] <= user_requirements["max_weight_kg"]) &
    (inventory["ram_gb"] >= user_requirements["min_ram_gb"]) &
    (inventory["storage_gb"] >= user_requirements["min_storage_gb"]) &
    (inventory["battery_hours"] >= user_requirements["min_battery_hours"])
]
```
